## Understanding DataSets

```python
import pandas as pd
url = "https://raw.githubusercontent.com/PratiikJagtap/Bitcoin-Market-Sentiment-Analysis-System/main/csv_files/historical_data.csv"
trades_df = pd.read_csv(url)
trades_df
```

```python
url2 = "https://raw.githubusercontent.com/PratiikJagtap/Bitcoin-Market-Sentiment-Analysis-System/main/csv_files/fear_greed_index.csv"
sentiment_df = pd.read_csv(url2)
sentiment_df
```

```python
# Encoding categorical columns
# converting into numerical
from sklearn.preprocessing import LabelEncoder

cat_cols = ["Side", "Direction", "Crossed", "Coin"]

le = LabelEncoder()
for col in cat_cols:
    trades_df[col] = le.fit_transform(trades_df[col])
```

```python
trades_df.head()
```

```python
trades_df["Timestamp IST"] = pd.to_datetime(
    trades_df["Timestamp IST"],
    format="%d-%m-%Y %H:%M"
)

# extracting date from Timestamp IST
trades_df["date"] = trades_df["Timestamp IST"].dt.date
trades_df["date"] = pd.to_datetime(trades_df["date"])
# this made new column in trade_df
```

```python
trades_df.head()
```

```python
daily_pnl_per_trader = (trades_df.groupby(["date", "Account"])["Closed PnL"].sum().reset_index(name="daily_pnl"))
daily_pnl_per_trader
```

```python
len(trades_df['Account'].unique())
```

```python
len(trades_df['date'].unique())
```

```python
trades_df["is_win"] = trades_df["Closed PnL"] > 0

daily_win_rate = (trades_df.groupby("date")["is_win"].mean().reset_index(name="win_rate"))
daily_win_rate
```

```python
avg_trade_size = (trades_df.groupby("date").agg(avg_tokens=("Size Tokens", "mean"), avg_usd=("Size USD", "mean")).reset_index())
avg_trade_size
# Average trade size (tokens & USD)
```

```python
leverage_dist = (trades_df.groupby("date")["Start Position"].mean().reset_index(name="avg_leverage"))
leverage_dist
```

```python
trades_per_day = (trades_df.groupby("date").size().reset_index(name="trade_count"))
trades_per_day
```

```python
len(trades_df.loc[trades_df['date'] == '2025-04-30', 'date'])
# Checking how trade count appear
```

```python
long_short_ratio = (trades_df.groupby("date")["Direction"].value_counts(normalize=True).unstack(fill_value=0).reset_index())
long_short_ratio
```

```python
# droping unwanted columns

id_cols = [
    "Account",
    "Transaction Hash",
    "Order ID",
    "Trade ID"
]

trades_df = trades_df.drop(columns=id_cols)
```

```python
# Aggregate trading data per date
# here we are converting many trades one day into one row per day
daily_trades = (
    trades_df
    .groupby("date")
    .agg(
        avg_price=("Execution Price", "mean"),
        total_volume=("Size Tokens", "sum"),
        total_usd=("Size USD", "sum"),
        trade_count=("Execution Price", "count"),
        total_fee=("Fee", "sum"),
        avg_position=("Start Position", "mean"),
        avg_pnl=("Closed PnL", "mean"),
        side_mean=("Side", "mean"),
        direction_mean=("Direction", "mean"),
        crossed_mean=("Crossed", "mean")
    )
    .reset_index()
)
```

```python
sentiment_df["date"] = pd.to_datetime(sentiment_df["date"])
```

```python
final_df = pd.merge(
    daily_trades,
    sentiment_df[["date", "value", "classification"]],
    on="date",
    how = "inner"
)
```

# EDA

```python
performance_summary = (
    final_df
    .groupby("classification")
    .agg(
        avg_pnl=("avg_pnl", "mean"),
        win_rate=("avg_pnl", lambda x: (x > 0).mean()),
        pnl_std=("avg_pnl", "std")
    )
)

performance_summary
```

### Observations:
#### Average PnL:
- When sentiment is Greed or Extreme Greed, traders make higher average profits.
- When sentiment is Fear or Extreme Fear, traders make lower average profits.

#### Win Rate:
- When sentiment is Extreme Greed, trades have a higher chance of being profitable.
- When sentiment is Neutral, fewer trades are profitable, but winning trades are larger.

#### PnL Standard Deviation:
- When sentiment is Neutral, trading risk and volatility are highest.
- When sentiment is Fear or Extreme Fear, trading behavior is more controlled but less profitable.
- When sentiment is Extreme Greed, profits increase but risk also increases.

```python
behavior_summary = (
    final_df
    .groupby("classification")
    .agg(
        avg_trades=("trade_count", "mean"),
        avg_leverage=("avg_position", "mean"),
        avg_volume=("total_volume", "mean"),
        long_bias=("direction_mean", "mean")
    )
)

behavior_summary
```

#### Observations
- When sentiment is Fear or Extreme Fear, traders take more trades.
- When sentiment is Greed or Extreme Greed, traders take fewer trades.
- During Fear and Neutral conditions, traders use higher leverage.
- During Greed, leverage usage is lower.
- In Extreme Greed, traders place bigger trades but less frequently.
- Long/short bias remains mostly the same across all sentiment conditions.

```python
# Segment 1: High vs Low Leverage Days

final_df["leverage_segment"] = (final_df["avg_position"] > final_df["avg_position"].median()).astype(int)
final_df.groupby("leverage_segment")["avg_pnl"].mean()
# 0 = low leverage & 1 = high leverage
```

- High-leverage days have a slightly higher average PnL
- Low-leverage days are a bit less profitable

```python
# Segment 2: Frequent vs Infrequent Trading Days
final_df["frequency_segment"] = (final_df["trade_count"] > final_df["trade_count"].median())
final_df.groupby("frequency_segment")["avg_pnl"].mean()
```

- When traders trade less frequently, the average PnL is higher.
- When traders trade more frequently, the average PnL is lower.

```python
# Segment 3: Consistent vs Inconsistent Performance
final_df["consistent_day"] = final_df["avg_pnl"] > 0
final_df.groupby("consistent_day")["trade_count"].mean()
```

- On consistent (profitable) days, traders take more trades.
- On inconsistent (loss-making) days, traders take fewer trades.
